```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import os
warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")
print('Libraries loaded ✓')
```

```python
#load dataset
df_processed = pd.read_csv("../data/processed_data/bank-additional-full-processed.csv")
        
df.head(1)
```

```python
df = df_processed.copy() #create working copy
```

```python
# New feature_ Was this client contacted in a previous campaign? 
df['was_contacted_before'] = (df['pdays'] != 999).astype(int)
```

```python
# emp.var.rate,euribor3m and, nr.employed variables are highly corerlated.
# create a composite feature (PCA)
# Confirm the correlations before treatment
macro_cols = ['emp.var.rate', 'euribor3m', 'nr.employed']
print("Correlations BEFORE treatment:")
print(df[macro_cols].corr().round(2))
```

```python
# Scale the three features before applying PCA
scaler_macro = StandardScaler()
macro_scaled = scaler_macro.fit_transform(df[macro_cols])
```

```python
# Apply PCA to reduce 3 correlated features into 1 component
pca_macro = PCA(n_components=1)
df['economic_index'] = pca_macro.fit_transform(macro_scaled)

# How much variance does this single component capture?
print(f"\nVariance explained by economic_index: "
      f"{pca_macro.explained_variance_ratio_[0]*100:.1f}%")
```

```python
# Verify the new feature looks sensible
print("\nNew feature 'economic_index' — sample statistics:")
print(df['economic_index'].describe().round(3))
```

```python
# Drop the original three correlated columns and pdays
df= df.drop(columns=['emp.var.rate', 'euribor3m', 'nr.employed'])
```

```python
# New feature_Is this customer in risk (default)
df['is_risk'] = df['default'].isin(['yes', 'unknown']).astype(int)
df = df.drop(columns=['default'])
```

```python
df['poutcome_success'] = (df['poutcome'] == 'success').astype(int)
print(df['poutcome_success'].value_counts())
df = df.drop(columns=['poutcome']) #drop original poutcome column
```

```python
df.head(1)
```

```python
df['is_cellular'] = (df['contact'] == 'cellular').astype(int)
df = df.drop(columns=['contact']) #drop original contact column
```

```python
# understand age distribution before deciding bins

print("Age Statistics:")
print(df['age'].describe().round(2))

print(f"\nMinimum age : {df['age'].min()}")
print(f"Maximum age : {df['age'].max()}")
print(f"Mean age    : {df['age'].mean():.1f}")
print(f"Median age  : {df['age'].median():.1f}")
```

```python
# Check the subscription rate of each age
age_rate = df.groupby('age')['y'].apply(
    lambda x: (x == 'yes').mean() * 100
).round(2)

plt.figure(figsize=(14, 5))
plt.plot(age_rate.index, age_rate.values, color='#0b5d4a', linewidth=2)
plt.axhline(y=df['y'].eq('yes').mean()*100, 
            color='red', linestyle='--', alpha=0.5, 
            label='Overall average')
plt.title("Subscription Rate by Exact Age", fontweight='bold')
plt.xlabel("Age")
plt.ylabel("Subscription Rate (%)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[17, 25, 60, 100],
    labels=['Students/early career', 'working_age', 'senior/retired']
)

df = df.drop(columns=['age']) #drop original age column
```
